```python
# Install necessary libraries (run once in Colab)
!pip install nltk scikit-learn

import nltk
import string
from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Download required NLTK resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

# Movie Review Text
text = """The movie was entertaining from beginning to end and featured excellent performances
by the lead actors. The visual effects and background music added to the overall
experience, making several scenes memorable. Although the story slowed down slightly
in the second half, the emotional ending made up for it. I would recommend this movie
to anyone who enjoys action, drama, and inspiring stories."""

# 1. Sentence Segmentation
sentences = sent_tokenize(text)
print("Sentence Segmentation:", sentences)

# 2. Tokenization
tokens = word_tokenize(text)
print("\nTokenization:", tokens)

# 3. Lowercasing
tokens_lower = [w.lower() for w in tokens]
print("\nLowercasing:", tokens_lower)

# 4. Remove punctuation & numbers
tokens_clean = [w for w in tokens_lower if w.isalpha()]
print("\nClean Tokens:", tokens_clean)

# 5. Remove Stopwords
stop_words = set(stopwords.words('english'))
tokens_no_stop = [w for w in tokens_clean if w not in stop_words]
print("\nWithout Stopwords:", tokens_no_stop)

# 6. Stemming
stemmer = PorterStemmer()
tokens_stemmed = [stemmer.stem(w) for w in tokens_no_stop]
print("\nStemming:", tokens_stemmed)

# 7. Lemmatization
lemmatizer = WordNetLemmatizer()
tokens_lemmatized = [lemmatizer.lemmatize(w) for w in tokens_no_stop]
print("\nLemmatization:", tokens_lemmatized)

# 8. Bag of Words
vectorizer = CountVectorizer()
bow = vectorizer.fit_transform([text])
print("\nBag of Words Vocabulary:", vectorizer.get_feature_names_out())
print("BoW Matrix:\n", bow.toarray())

# 9. TF-IDF
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform([text])
print("\nTF-IDF Vocabulary:", tfidf.get_feature_names_out())
print("TF-IDF Matrix:\n", tfidf_matrix.toarray())

# 10. Sentiment Analysis (Simple Rule-Based)
positive_words = ["entertaining","excellent","memorable","recommend","enjoys","inspiring"]
negative_words = ["slowed","slightly"]

pos_count = sum([1 for w in tokens_lower if w in positive_words])
neg_count = sum([1 for w in tokens_lower if w in negative_words])

if pos_count > neg_count:
    sentiment = "Positive"
elif neg_count > pos_count:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

print("\nOverall Sentiment:", sentiment)
```
